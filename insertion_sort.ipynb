```python
def insertion_sort(lista):
    n = len(lista)
    for i in range(1, n):
        clave = lista[i]
        j = i - 1
        while j >= 0 and lista[j] > clave:
            lista[j + 1] = lista[j]
            j = j - 1
        lista[j + 1] = clave
    return lista

lista_ordenada = [1, 2, 3, 4, 5]
lista_invertida = [9, 8, 7, 6, 5]
lista_repetidos = [4, 1, 4, 2, 1]
lista_un_elemento = [42]

print(insertion_sort(lista_ordenada))
print(insertion_sort(lista_invertida))
print(insertion_sort(lista_repetidos))
print(insertion_sort(lista_un_elemento))
```
